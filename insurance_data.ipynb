#Task 3 — Center and Spread for One Column

```python
print("mean:", df['bmi'].mean())
print("median:", df['bmi'].median())
print("minimum:", df['bmi'].min())
print("maximum:", df['bmi'].max())
print("standard deviation:", df['bmi'].std())
```

1. The mean and median is close. The shape of the distrubution will look like a bell.

# Task 4

```python
df['bmi'].plot(kind='hist', bins=20)
plt.title('Distribution of BMI')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()
```

The distribution is centered around the mean and median, and the spread is predictable and seems to be like a normal distribution. There are very few outliers
